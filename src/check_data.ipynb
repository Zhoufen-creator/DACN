```python
import pandas as pd 
import matplotlib.pyplot as plt

df = pd.read_csv("D:\DACN\dataset\creditcard.csv")
counts = df["Class"].value_counts().sort_index()

labels = ["Normal (0)", "Fraud (1)"]

plt.pie(counts, startangle=90)

plt.legend(
    [f"{label}: {count}" for label, count in zip(labels, counts)],
    loc="upper center",
    bbox_to_anchor=(0.5, -0.05),
    ncol=2
)


plt.title("Credit Card Fraud Distribution")
plt.axis("equal")
plt.show()
```
